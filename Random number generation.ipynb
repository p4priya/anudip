### RANDOM NUMBER GENERATION

```python
import numpy as np
```

```python
#Uniform distribution
uniform=np.random.rand(3,3)
print("Uniform distribution:\n",uniform)
```

```python
#normal distribution
normal=np.random.randn(3,3)
print("normal distribution:\n",normal)
```
